```python
# Импорт необходимых библиотек
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal

%matplotlib inline
```

## Математическое описание

### Амплитудная модуляция (AM)
Амплитудная модуляция изменяет амплитуду несущего сигнала в зависимости от модулирующего сигнала. Формула AM-сигнала:
\[
s_{AM}(t) = [A_c + m \cdot s_m(t)] \cdot \cos(2\pi f_c t),
\]
где:
- \( A_c \) — амплитуда несущего сигнала,
- \( m \) — коэффициент модуляции,
- \( s_m(t) = A_m \cdot \cos(2\pi f_m t) \) — модулирующий сигнал,
- \( f_c \) — частота несущего сигнала,
- \( f_m \) — частота модулирующего сигнала.

### Частотная модуляция (FM)
Частотная модуляция изменяет частоту несущего сигнала в зависимости от модулирующего сигнала. Формула FM-сигнала:
\[
s_{FM}(t) = A_c \cdot \cos\left(2\pi f_c t + k_f \int_0^t s_m(\tau) d\tau\right),
\]
где:
- \( k_f \) — коэффициент частотной девиации,
- \( \int_0^t s_m(\tau) d\tau \) — интеграл модулирующего сигнала.

```python
# Параметры сигналов
f_c = 100  # Частота несущего сигнала (Гц)
f_m = 10   # Частота модулирующего сигнала (Гц)
A_c = 1.0  # Амплитуда несущего сигнала
A_m = 0.5  # Амплитуда модулирующего сигнала
m = 0.5    # Коэффициент модуляции для AM
k_f = 50   # Коэффициент частотной девиации для FM
f_s = 10000  # Частота дискретизации (Гц)
t = np.linspace(0, 0.2, int(0.2 * f_s))  # Временной массив

# Генерация несущего и модулирующего сигналов
carrier = A_c * np.cos(2 * np.pi * f_c * t)  # Несущий сигнал
modulating = A_m * np.cos(2 * np.pi * f_m * t)  # Модулирующий сигнал

# Амплитудная модуляция
am_signal = (A_c + m * modulating) * np.cos(2 * np.pi * f_c * t)

# Частотная модуляция
# Интеграл модулирующего сигнала для FM
phase = k_f * np.cumsum(modulating) / f_s
fm_signal = A_c * np.cos(2 * np.pi * f_c * t + phase)
```

## Пояснение к коду

1. **Параметры**:
   - Заданы частоты \( f_c = 100 \, \text{Гц} \), \( f_m = 10 \, \text{Гц} \), амплитуды \( A_c = 1 \), \( A_m = 0.5 \).
   - Коэффициент модуляции \( m = 0.5 \) для AM и \( k_f = 50 \) для FM.
   - Временной массив \( t \) создается с частотой дискретизации \( f_s = 10000 \, \text{Гц} \).

2. **Сигналы**:
   - Несущий сигнал: \( A_c \cdot \cos(2\pi f_c t) \).
   - Модулирующий сигнал: \( A_m \cdot \cos(2\pi f_m t) \).
   - AM-сигнал: вычисляется по формуле \( [A_c + m \cdot s_m(t)] \cdot \cos(2\pi f_c t) \).
   - FM-сигнал: вычисляется с использованием интеграла модулирующего сигнала, реализованного через `np.cumsum`.

3. **Частота дискретизации**:
   - \( f_s = 10000 \, \text{Гц} \) выбрана достаточно высокой, чтобы точно представить сигналы.

```python
# Создание графиков
plt.figure(figsize=(12, 8))

# Несущий сигнал
plt.subplot(3, 2, 1)
plt.plot(t, carrier, color='blue')
plt.title('Несущий сигнал')
plt.xlabel('Время (с)')
plt.ylabel('Амплитуда')
plt.grid(True)

# Модулирующий сигнал
plt.subplot(3, 2, 2)
plt.plot(t, modulating, color='green')
plt.title('Модулирующий сигнал')
plt.xlabel('Время (с)')
plt.ylabel('Амплитуда')
plt.grid(True)

# AM сигнал
plt.subplot(3, 2, 3)
plt.plot(t, am_signal, color='red')
plt.title('Амплитудно-модулированный сигнал')
plt.xlabel('Время (с)')
plt.ylabel('Амплитуда')
plt.grid(True)

# FM сигнал
plt.subplot(3, 2, 4)
plt.plot(t, fm_signal, color='purple')
plt.title('Частотно-модулированный сигнал')
plt.xlabel('Время (с)')
plt.ylabel('Амплитуда')
plt.grid(True)

plt.tight_layout()
plt.show()
```

## Анализ результатов

- **Несущий сигнал**: Синусоида с частотой 100 Гц и амплитудой 1.
- **Модулирующий сигнал**: Синусоида с частотой 10 Гц и амплитудой 0.5.
- **AM-сигнал**: Амплитуда несущего сигнала изменяется в соответствии с модулирующим сигналом. Видно, что огибающая сигнала повторяет форму модулирующего сигнала.
- **FM-сигнал**: Частота несущего сигнала изменяется в зависимости от модулирующего сигнала, что проявляется в изменении плотности синусоидальных колебаний на графике.

Графики наглядно демонстрируют различия между AM и FM модуляцией.
